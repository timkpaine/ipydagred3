# ipydagred3

Run the cells from top to bottom to draw a network, update its graph, and handle clicks in Python.

```python
from ipywidgets import Label, VBox

from ipydagred3 import DagreD3Widget, Graph
```

## Network graph

Draw TCP connection states, then update their colors and add nodes and edges.

```python
network = Graph()
```

```python
states = ["CLOSED", "LISTEN", "SYN RCVD", "SYN SENT", "ESTAB", "FINWAIT-1", "CLOSE WAIT", "FINWAIT-2", "CLOSING", "LAST-ACK", "TIME WAIT"]
```

```python
for state in states:
    network.setNode(state, tooltip="tooltip1 of " + state)
```

```python
network.setEdge("CLOSED", "LISTEN", label="open")
network.setEdge("LISTEN", "SYN RCVD", label="rcv SYN")
network.setEdge("LISTEN", "SYN SENT", label="send")
network.setEdge("LISTEN", "CLOSED", label="close")
network.setEdge("SYN RCVD", "FINWAIT-1", label="close")
network.setEdge("SYN RCVD", "ESTAB", label="rcv ACK of SYN")
network.setEdge("SYN SENT", "SYN RCVD", label="rcv SYN")
network.setEdge("SYN SENT", "ESTAB", label="rcv SYN, ACK")
network.setEdge("SYN SENT", "CLOSED", label="close")
network.setEdge("ESTAB", "FINWAIT-1", label="close")
network.setEdge("ESTAB", "CLOSE WAIT", label="rcv FIN")
network.setEdge("FINWAIT-1", "FINWAIT-2", label="rcv ACK of FIN")
network.setEdge("FINWAIT-1", "CLOSING", label="rcv FIN")
network.setEdge("CLOSE WAIT", "LAST-ACK", label="close")
network.setEdge("FINWAIT-2", "TIME WAIT", label="rcv FIN")
network.setEdge("CLOSING", "TIME WAIT", label="rcv ACK of FIN")
network.setEdge("LAST-ACK", "CLOSED", label="rcv ACK of FIN")
network.setEdge("TIME WAIT", "CLOSED", label="timeout=2MSL");
```

```python
network_widget = DagreD3Widget(graph=network)
network_widget
```

```python
# Add some custom colors based on state
network.setNode("CLOSED", style="fill: #f77")
network.setNode("ESTAB", style="fill: #7f7")
```

```python
# Add new nodes
network.setNode("test")
network.setNode("test2")
```

```python
# Add new edges
network.setEdge("test", "test2")
```

## Tooltips and click events

Hover over a node or edge to see its tooltip. Click a node or edge to update the label below.

```python
tooltips = Graph(multigraph=True)
n = tooltips.setNode("a", shape="diamond")

tooltips.setNode("b", tooltip="this is a test")
e = tooltips.setEdge("a", "b", label="test", labelpos="r", tooltip="a->b")
tooltip_widget = DagreD3Widget(graph=tooltips)
```

```python
def on_click(graph, clicked):
    tooltip_box.children[-1].value = f"last clicked: {clicked}"


tooltip_widget.on_click(on_click)

tooltip_box = VBox([tooltip_widget, Label()])
tooltip_box
```
